```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
df = pd.read_csv("Arrest_Data_from_2010_to_2019.csv")
```

```python
data = df
data['Arrest Date'] = pd.to_datetime(data['Arrest Date'], format='%m/%d/%Y')
start_date = pd.to_datetime('01/01/2010', format='%m/%d/%Y')
end_date = pd.to_datetime('12/31/2018', format='%m/%d/%Y')
filtered_data = data.loc[(data['Arrest Date'] >= start_date) & (data['Arrest Date'] <= end_date) & ((data['Arrest Type Code'] == 'F') | (data['Arrest Type Code'] == 'M'))].copy()
filtered_data['Year'] = filtered_data['Arrest Date'].dt.year
arrest_by_year = filtered_data.groupby('Year').size()
```

```python
arrest_by_year.index = [int(i) for i in arrest_by_year.index]
x = np.array(arrest_by_year.index)
y = np.array(arrest_by_year.values)
mean_x = x.mean()
mean_y = y.mean()
mean_y
```

```python
Sxx = 0
Sxy = 0
for i in range(len(x)):
    Sxx += (x[i]-mean_x)**2
    Sxy += (y[i] - mean_y)*(x[i]-mean_x)
b = Sxy/Sxx
a = mean_y - b*mean_x
y = a + b*2019
y
```
